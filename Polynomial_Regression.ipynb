```python
#AP24110010798
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
#AP24110010798
# Dataset
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([1, 4, 9, 16, 25])
#AP24110010798
# Convert X into polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
#AP24110010798
# Train the model
model = LinearRegression()
model.fit(X_poly, Y)
#AP24110010798
# Prediction
Y_pred = model.predict(X_poly)
#AP24110010798
# Display results
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)
print("Predicted values:", Y_pred)
#AP24110010798
# Predict for X=6
new_X = np.array([[6]])
new_X_poly = poly.transform(new_X)
print("Prediction for X=6:", model.predict(new_X_poly)[0])
#AP24110010798
# Plot graph
plt.scatter(X, Y, color="blue", label="Actual Data")
plt.plot(X, Y_pred, color="red", label="Polynomial Curve")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()

```

Polynomial Regression

Models a nonlinear relationship between variables using polynomial terms such as \(X^2\) and \(X^3\).

Formula: \(Y = b_0 + b_1X + b_2X^2\)

Example: Predicting population growth.
